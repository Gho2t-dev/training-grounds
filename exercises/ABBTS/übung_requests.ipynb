# Eine übung für Requests


```python
import requests

# einfache wetter api gefunden auf google
url = "https://api.open-meteo.com/v1/forecast"

# paramtere für baden festlegen damit die api weiss von wo wir das wetter wissen wollen
params = {
    "latitude": 47.47,      # Baden, Schweiz
    "longitude": 8.31,
    "current_weather": True # Wir wollen das aktuelle Wetter
}

# die eigentliche Request
response = requests.get(url, params=params)

# Prüfen, ob alles geklappt hat (Status 200 = OK)
if response.status_code == 200:
    # Die Antwort kommt als JSON, wir wandeln sie in ein Python-Dictionary um
    daten = response.json()
    
    # Wir holen uns die Temperatur aus dem Dictionary
    temperatur = daten["current_weather"]["temperature"]
    
    print(f"Aktuelle Temperatur: {temperatur}°C")
else:
    print("Etwas ist schiefgelaufen:", response.status_code)
```
