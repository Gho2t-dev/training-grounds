#Szenario
In einer Werkstatt werden Aufträge und verwendete Bauteile analysiert.


```python
# Gegebene Daten

# Beispiel erster Eintrag: Autragsnummer 101, Bauteil: Motor, Verwendete Bauteile 2 Schrauben, 1 Mutter
auftraege = [
    (101, "Motor", ["Schraube", "Mutter", "Schraube"]),
    (102, "Sensor", ["Sensor", "Kabel"]),
    (103, "Motor", ["Schraube", "Dichtung"]),
    (104, "Relais", ["Kontakt", "Spule"]),
]

lagerbestand = {
    "Schraube": 100,
    "Mutter": 80,
    "Sensor": 20,
    "Kabel": 50,
    "Dichtung": 10,
    "Kontakt": 5,
    "Spule": 2
}
```



1.  Verschachtelte Schleifen
Durchlaufe alle Aufträge und alle Bauteile darin.
→ Zähle, wie oft jedes Bauteil insgesamt verwendet wird.
→ Speichere in einem Dictionary.

---

2. if / else Logik
Vergleiche Verbrauch mit Lagerbestand.

Gib aus:
-  OK (genug Bestand)
-  Nachbestellen (wenn Verbrauch > Bestand)

---
3. Sets
   
- Erstelle eine Menge aller verwendeten Bauteile
- Erstelle eine Menge aller kritischen Bauteile
- Bestimme die Schnittmenge

---
4. Datentypenwahl
Beantworte:
- Warum Liste für Aufträge?
- Warum Tuple für Auftrag?
- Warum Dict für Lager?
- Warum Set für Analyse?

---
5. Bonus
- Finde Bauteile, die in mehreren Aufträgen vorkommen
- Nutze verschachtelte Schleifen + Set


```python
# for i in auftraege[0][2]:
    #print(i) # Test wie ich verschachtelte listen und tuppel entpacke

items = {}
n = 0
for a in auftraege:
    for i in auftraege[n][2]:
        if i in items.keys():
            items[i] += 1
        else:
            items[i] = 1
    n += 1

print(items)
```
