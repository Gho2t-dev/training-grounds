# 1. Variablen
In anderen Sprachen (C, C++, Java, ...) muss einer Variablen, vor der ersten Verwendung, ein Datentyp zugeordnet werden. Dieser ist für die Laufzeit des Programms fix und kann nicht mehr verändert werden. In Python ist dies nicht der Fall, wie das nachfolgende Beispiel zeigt.

```python
a = 3
print('Wert von a = ', a)
print('a ist vom Datentyp: ', type(a))

a = 3.0
print('Wert von a = ', a)
print('a ist vom Datentyp: ', type(a))

a = 'Ich bin ein String'
print('Wert von a = ', a)
print('a ist vom Datentyp: ', type(a))

a = {'Rot': 'Gruen', 'Schwarz': 'Weiss'}
print('Wert von a = ', a)
print('a ist vom Datentyp: ', type(a))
```

Wie im obigen Beispiel ersichtlich, ändert die Variable *a* nicht nur ihren Wert, sondern ebenfalls ihren Datentyp. Mehr zu den Datentypen im Kapitel 2. Zusätzlich zum Datentyp einer Variablen weisen diese eine Identität auf. Diese ist eine eindeutige Zuweisung einer Variablen zu einem Objekt (Speicherort). Falls zwei Variablen gleich sind (nicht nur vom Wert) weisen sie dieselbe Identität auf.

```python
a = 123.0
print('Wert von a = ', a, 'und weist die Indentität: ', id(a), 'auf')
b = 123.0
# a und b sind vom Wert und Datentyp identisch
print('a und b sind vom Wert her gleich? ', a == b)
print('a und b sind vom Datentyp her gleich? ', type(a) == type(b))
print('a und b weisen dieselbe Indentität auf? ', id(a) == id(b))
b = a
print('a und b weisen dieselbe Indentität auf? ', id(a) == id(b)) 
```

## 1.2 Konvention für Variablennamen
Variablennamen müssen zwingend mit einem Buchstaben oder einem Unterstrich `_` beginnen und sind case-sensitive (Unterscheidung Gross-/Kleinschreibung). Weiter gilt in der Python-Community, Kleinschreibung bei Variablennamen und falls der Variablenname aus mehreren Wörtnern besteht, so werden diese mit `_` separiert. Zusätzlich sollten möglichst sprechende Variablennamen verwendet werden, wie z.B. *maximale_beschleunigung*.

### 1.2.1 Schlüsselwörter
Es existieren in Python 3, 30 reservierte Schlüsselwörter, welche nicht für die Bezeichnung von Variablen benutzt werden dürfen. <br> <br>
*and, as, assert, break, class, continue, def, del, elif, else, except, finally, for, from, global, if, import, in, is, lambda, nonlocal, not, or, pass, raise, return, try, while,  with, yield* <br> <br>
Falls eines dieser Schlüsselwörter als Variablennamen verwendet wird, wird ein Syntax-Fehler erzeugt.
