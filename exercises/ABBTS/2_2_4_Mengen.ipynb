### 2.4.1 Mengen (set)
Eine Menge (wie aus der Mathematik bekannt) enthält eine ungeordnete Sammlung von einmaligen und unveränderlichen Elementen. D.h. ein Element kann in einer Menge nicht mehrmals vorkommen, was z.B. bei Listen und Tupeln möglich ist. Die Unveränderbarkeit bezieht sich nur auf die Elemente, aber nicht auf die Menge als solches, d.h. der Inhalt der Menge ist veränderlich (Anzahl Elemente). Falls keine Veränderungen an einer Menge erlaubt sind, können sogenannte *frozensets* erzeugt werden.

```python
studenten = {'Student X', 'Student Y', 'Student Z', 'Student W'}
print(studenten)
# Zugehörigkeitsfunktion in
print('Ist Student X Teil der Menge studenten? ', 'Student X' in studenten)
noten = frozenset((4.5, 3.5, 5.2, 5.7))  # unveränderliche Menge
```

### 2.4.2 Operationen auf Mengen
1. *add()*
2. *copy()*
3. *clear()*
4. *difference()*
5. *difference_update()*
6. *discard()*
7. *remove()*
8. *intersection()*
9. *isdisjoint()*
10. *issubset()*
11. *issuperset()*
12. *pop()*

```python
# add()
''' hiermit wird ein neues Element einer Menge hinzugefügt. Falls das Element bereits
    existiert geschieht nichts. '''
studenten = {'Student X', 'Student Y', 'Student Z', 'Student W'}
studenten.add('Student K')
print(studenten)
studenten.add('Student X')  # falls Element bereits vorhanden ist, geschieht nichts
print(studenten)
# frozensets sind unveränderlich, betreffend Inhalt der Menge
noten = frozenset((4.5, 3.5, 5.2, 5.7))  # unveränderliche Menge
noten.add(4.2)
```

```python
# copy()
''' erzeugt eine Kopie einer Menge. ''' 
studenten = {'Student X', 'Student Y', 'Student Z', 'Student W'}
studenten_2 = studenten.copy()  # Kopiert ein Dictionary, es wird ein neues Objekt erzeugt
studenten_2.add('Student A')
print(studenten)
print(studenten_2)
```

```python
# clear()
''' löscht die Elemente eines Dictionaries. ''' 
studenten = {'Student X', 'Student Y', 'Student Z', 'Student W'}
studenten.clear()  # Löscht die Elemente, die Menge aber nicht
print(studenten)
print(type(studenten))
```

```python
# difference()
''' liefert die Differenz von zwei oder mehreren Mengen zurück. '''
studenten_1 = {'Student X', 'Student Y', 'Student Z', 'Student W'}
studenten_2 = {'Student X', 'Student Z'}
studenten_3 = {'Student Y', 'Student Z'}

print(studenten_1.difference(studenten_2))
print(studenten_1.difference(studenten_2).difference(studenten_3))
# anstelle von 'difference', funktioniert auch '-'
print(studenten_1 - studenten_2)
print(studenten_1 - studenten_2 - studenten_3)
```

```python
# x.difference_update(s)
''' entfernt alle Elemente von s aus x -> x = x-y '''
studenten_1 = {'Student X', 'Student Y', 'Student Z', 'Student W'}
studenten_2 = {'Student X', 'Student Z'}
studenten_1.difference_update(studenten_2)
print(studenten_1)
```

```python
# x.discard(e)
''' das Elemente e wird aus x entfernt, falls es enthalten ist.
    Ist das Element nicht vorhanden geschieht nichts.''' 
studenten = {'Student X', 'Student Y', 'Student Z', 'Student W'}
studenten.discard('Student Z')
print(studenten)
```

```python
# x.intersection(s)
''' liefert die Schnittmenge von s und x zurück -> & '''
studenten_1 = {'Student X', 'Student Y', 'Student Z', 'Student W'}
studenten_2 = {'Student X', 'Student Z', 'Student A'}
print(studenten_1.intersection(studenten_2))
# Alternativ:
print(studenten_1 & studenten_2)
```

```python
# m1.isdisjoint(m2)
''' liefert 'True' zurück, falls zwei Mengen (m1, m2) elementfremd sind. ''' 
studenten_1 = {'Student X', 'Student Y', 'Student Z', 'Student W'}
studenten_2 = {'Student X', 'Student Z', 'Student A'}
studenten_3 = {'Student A'}
print(studenten_1.isdisjoint(studenten_2))
print(studenten_1.isdisjoint(studenten_3))
```

```python
# x.issubset(y)
''' liefert 'True' zurück, falls x eine echte Teilmenge von y ist.
    y darf nur Elemente aus x enthalten. '''
studenten_1 = {'Student X', 'Student Y', 'Student Z', 'Student W'}
studenten_2 = {'Student X', 'Student Z'}
studenten_2.issubset(studenten_1)
```

```python
# x.issuperset(y)
''' liefert 'True' zurück, falls x eine Obermenge von y ist. '''
studenten_1 = {'Student X', 'Student Y', 'Student Z', 'Student W'}
studenten_2 = {'Student X', 'Student Z'}
studenten_1.issuperset(studenten_2)
```

```python
# pop()
''' liefert ein beliebiges Element der Menge zurück und entfernt dieses aus der Menge.
    Ist die Menge leer wird ein Fehler erzeugt.'''
studenten = {'Student X', 'Student Y', 'Student Z', 'Student W'}
print(studenten.pop())
print(studenten)
print(studenten.pop())
print(studenten)
```
