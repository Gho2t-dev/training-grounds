# 5. Dateien lesen und schreiben
Um Daten aus einer Datei zu lesen oder auch zu schreiben, muss die entsprechende Datei zuerst mit der open-Funktion geöffnet werden. Damit wird ein Dateiobjekt erzeugt und sie liefert eine Referenz auf dieses Objekt als Ergebnis zurück.

## 5.1 Lesen einer Datei

```python
dateiname = 'ZUSATZ/messung.txt'
fobj = open(dateiname, 'r')
print(fobj)
for linie in fobj:
    # print(linie)
    print(linie.rstrip())  # rstrip: entfernt allfällige Leerzeichen und Newlines vom rechten Rand
fobj.close()
```

## 5.2 Schreiben einer Datei

```python
dateiname_in = 'ZUSATZ/messung_in.txt'
dateiname_out = 'ZUSATZ/messung_out.txt'
fobj_in = open(dateiname_in, 'r')
fobj_out = open(dateiname_out, 'w')

anzahl_zeilen = 0

for linie in fobj_in:
    anzahl_zeilen += 1
    zeile = '{:.0f}\t'.format(anzahl_zeilen)  # Zeilennummerierung
    zeile += linie.rstrip() + '\n'
    fobj_out.write(zeile)

fobj_in.close()
fobj_out.close()
```

## 5.3 Die Methoden read() und readlines()
In den oberen Beispielen wurde die Datei Zeile für Zeile gelesen bzw. geschrieben. Häufig möchte man aber den Inhalt einer Datei in eine komplette Datenstruktur einlesen. Auf diese Weise kann die Datei schnell wieder geschlossen und man arbeitet nur noch auf der Datenstruktur und nicht mehr mit der Datei.

```python
dateiname = 'ZUSATZ/messung.txt'
messung = open(dateiname).readlines()  # Zeilenweises Lesen der Datei
print(len(messung))  # Anzahl Zeilen, jede Zeile ist ein einzelner String
print(messung[0][0:20])

messung = open(dateiname).read()  # komplette Datei in einen String einlesen
print(len(messung))  # ein einziger String
print(messung[:20])
```

## 5.4 with-Anweisung
Benutzt man die with-Anweisung mit einer *open*, so wird die Datei nach Ablauf des Blockes oder bei Auftreten einer Ausnahme automatisch geschlossen. Damit ist sichergestellt, dass die Datei immer geschlossen wird, auch ohne ein explizites *close*. <br>
Mit *with* können auch mehrere Dateien geöffnet werden.

```python
dateiname_in = 'Zusatz/messung_in.txt'
dateiname_out = 'Zusatz/messung_out_with.txt'

with open(dateiname_in, 'r') as fobj_in, open(dateiname_out, 'w') as fobj_out:
    anzahl_zeilen = 0
    for linie in fobj_in:
        anzahl_zeilen += 1
        zeile = '{:.0f}\t'.format(anzahl_zeilen)
        zeile += linie.rstrip() + '\n'
        fobj_out.write(zeile)
```
