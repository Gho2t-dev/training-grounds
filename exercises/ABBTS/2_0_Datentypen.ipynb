# 2. Datentypen
Es existieren verschiedene Arten von Datentypen.
- <a href="2_1_Numerische_Datentypen.ipynb" target="_blank">Numerische Datentypen:</a>
    - Ganzzahl, Integer (int)
    - Fliesskommazahl, Floating-point (float)
    - Boolsche Werte (bool)
    - komplexe Zahl (complex)

- <a href="2_2_Sequentielle_Datentypen.ipynb" target="_blank">Sequentielle Datentypen:</a>
    - <a href="2_2_1_Strings.ipynb" target="_blank">Zeichenketten, Strings (str)</a>
    - <a href="2_2_2_Listen_Tupel.ipynb" target="_blank">Listen (list)</a>
    - <a href="2_2_2_Listen_Tupel.ipynb" target="_blank">Tupel (tuple)</a>

- Datentyp "Mapping":
    - <a href="2_2_3_Dictionaries.ipynb" target="_blank">Dictionaries (dict)</a>

- Datentyp "Collection":
    - <a href="2_2_4_Mengen.ipynb" target="_blank">Set (Mengen)</a>

```python
x = 3  # Ganzzahl, Integer
y = 3.45  # Fliesskommazahl, Floating-Point
z = True  # Boolscher Wert
w = 3 + 5j  # Komplexe Zahl

a = 'Das ist ein String'  # Zeichenkette String
b = [3, 4.67, 'Liste']  # Liste
c = (3, 4.67, 'Tupel')  # Tupel

d = {'Kp': 2.45, 'Ti': 0.34}  # Dictionary

print('x: ', x, type(x))
print('y: ', y, type(y))
print('z: ', z, type(z))
print('w: ', w, type(w))

print('a: ', a, type(a))
print('b: ', b, type(b))
print('c: ', c, type(c))
print('d: ', d, type(d))
```

## Datentyp ermitteln und umwandeln
Wie bereits in den obigen Beispielen ersichtlich kann der Datentyp über die Funktion *type()* ermittelt werden. Eine weitere Möglichkeit dies zu tun, erlaubt die eingebaute Funktion *isinstance()*, welche einen Wahrheitswert *True* oder *False* zurückgibt.

```python
x = 3.45
isinstance(x, float)
```

Der Datentyp kann folgendermassen verändert werden.

```python
x = 3.45
y = int(x)
print('Datentyp von x und y lauten: ', type(x), type(y))
y = float(y)
print('Datentyp y lautet: ', y, type(y))
```
