## 2.1 Numerische Datentypen
### 2.1.1 Ganze Zahlen
Ganze Zahlen $ -3, -2, -1, 0, 1, 2, 3 $ werden als Integer bezeichnet und sind in Python, im Gegensatz zu anderen Programmiersprachen, nicht durch einen Maximalwert begrenzt. Es können auch Binär, Oktal und Hexadezimalzahlen verwendet werden.

```python
x = 0b1001  # binär (2er-Zahlensystem)
y = 0o1001  # oktal (8er-Zahlensystem)
z = 0x1001  # hexadezimal (16er-Zahlensystem)
print(x, y, z)
print(type(x), type(y), type(z))
```

Mit den Funktionen bin, oct und hex verwandelt man die Integer-Zahl in einen String.

```python
x = bin(21)
y = oct(21)
z = hex(21)
print(x, y, z)
print(type(x), type(y), type(z))
```

### 2.1.2 Fliesskommazahlen
Fliesskommazahlen entsprechen dem Datentyp float in Python. Hier muss immer der Dezimalpunkt verwendet werden. Ebenfalls sehr nützlich ist die Potenzschreibweise, welche ebenfalls benutzt werden kann.

```python
x = 42.3456
y = 1.45e2
print('x = ', x)
print('Datentyp von x lautet: ', type(x))
print('y = ', y)
print('Datentyp von x lautet: ', type(y))
```

### 2.1.3 Boolsche Werte
Dieser Datentyp kennt nur zwei Zustände *True* und *False*. Die zugehörigen Operatoren lauten: *not*, *and* und *or*

```python
x = True
y = not x
z = x or y
print('x = ', x)
print('Datentyp von x lautet: ', type(x))
print('y = ', y)
print('z = ', z)
```

### 2.1.4 Komplexe Zahlen
Im Unterschied zu den meisten anderen Sprachen, bietet Python den Datentyp *complex*. Wie in der Elektrotechnik, nutzt Python j für die imaginäre Einheit. <br>
Dieser Datentyp besitzt zwei Attribute *real* und *imag* und eine Methode *conjugate()*. Auf diese beiden Begriffe Attribut und Methode werden wir im Rahmen der objektorientierten Programmierung detailliert eingehen. Jetzt genügt es zu wissen, dass dieser Datentyp diese Funktionalität hat.

```python
help('COMPLEX')
```

```python
import numpy as np

x = 2 + 6j
y = x.conjugate()  # konjugiert-komplex
print('x = ', x)
print('y = ', y)
print('Datentyp von x lautet: ', type(x))
print('Real- und Imaginärteil von x lauten: ', x.real, x.imag)
print('Datentyp von Real- und Imaginärteil lautet: ', type(x.real), type(x.imag))

x_betrag = np.round(abs(x), 2)
x_winkel = np.round(np.angle(x) / np.pi * 180, 2)
print('Der Betrag und der Winkel von x lauten: ', x_betrag, x_winkel)
```

## 2.2 Operatoren
Neben den bekannten mathematischen Operationen (+,-,`*`,/) existieren eine ganzzahlige Division, eine Restdivision, Betragsfunktion und Potenzieren.

```python
x = 9
y = 2

z_ganzzahl = x // y  # ganzzahlige Division
z_rest = x % y  # Modulo- oder Restdivision
print('Das Ergebnis der Ganzzahl-Division lautet: ', z_ganzzahl)
print('Das Ergebnis der Restdivision lautet: ', z_rest)

betrag = abs(-32.23)
print('Der Betrag von x lautet: ', betrag)

w = x ** y  # potenzieren 
print('x^y = ', w)
```
