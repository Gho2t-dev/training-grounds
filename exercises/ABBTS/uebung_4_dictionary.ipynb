 DICT – Sensorwerte

 Szenario
Mapping.

 Aufgaben
- Dict erstellen
- Ausgabe in einer for Schlaufe
- Ausbabe ausserhalb der for Schlaufe
- Ändern der Temp. auf 25
- Ausgabe

```python
sensors = {'temp': 20, 'pressure': 5}

for key, value in sensors.items():
    print(key, value)

print(sensors)

sensors['temp'] = 25
print(sensors)
```
