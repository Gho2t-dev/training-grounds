### 2.2.2 Strings (Zeichenketten) im Detail

Strings werden durch Listen repräsentiert und gehören somit zu den sequentiellen Datentypen. Damit kann ebenfalls das bekannte Indexieren resp. slicing angewendet werden.

```python
# Variablendeklaration
s = ' Hello World '
print(s)
```

```python
# Modifizieren von Strings
# https://www.w3schools.com/python/python_strings_methods.asp
s.upper()
```

```python
s.lower()
```

```python
s.title()
```

```python
s.strip()
```

```python
s.split()
```

```python
s.replace(' ', '_')
```

```python
s.find('o')  # suchen von links her beginnend
```

```python
s.rfind('o')  # suchen von rechts her beginnend
```

```python
# Concatenieren von Strings
s1 = 'Hello'
s2 = 'World'
```

```python
s1 + ' ' + s2
```

```python
# String Format
my_string = '{} danach {} und {} usw {}'
print(my_string.format('1', '2', '3', '4'))
```

```python
print('Die Temperatur beträgt innen {0:.2f} °C und aussen {1:.2f} °C.'
      .format(23.321, 34.456))
```

```python
print('Mein Name ist {0:3} und ich bin'
      .format('40'))
```

```python
# Erweiterung
tag = 8
text = f"Heute ist {tag}"
text
```

```python
# Escape Character \
s = 'Strings werden mittels \'\' eingefasst.'
print(s)
```

```python
# \n -> new line
s = 'Ich bin ein \nmehrzeiliger \nText.'
print(s)
```
