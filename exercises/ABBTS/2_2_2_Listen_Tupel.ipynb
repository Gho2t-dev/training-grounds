### 2.2.1 Listen und Tupel im Detail

Listen und Tupel beinhalten Elemente, welche über die Indizierung adressiert werden können. Ebenfalls kann die Anzahl der Elemente ermittelt werden. Nun werden die beiden Datentypen detaillierter betrachtet. Generell gilt alle Operationen, die zu keiner Veränderung des Inhalts führen, sind für Listen, wie auch Tupel definiert (z.B. *index()*, *count()*). Operationen, welche den Inhalt verändern, sind aufgrund der Unveränderbarkeit von Tupeln nur für Listen definiert (z.B. *pop()*, *append()*, *insert()*). <br>
Operationen:
- Elemente einer Liste anhängen
- Elemente aus einer Liste entfernen
- Inhalt einer Liste
- Index ermitteln
- Einfügen von Elementen
- Listen sortieren

```python
# Variablendeklaration
liste = [100, 3.9, 451, 55, 34]
tupel = (100, 3.9, 451, 55, 34)
zusaetzliches_element = 345.34
zusaetzliche_elemente_l = [45, 56, 10.9]
zusaetzliche_elemente_t = (45, 56, 10.9)
print(liste)
print(tupel)
```

```python
# am Ende einer Liste ein Element anfügen
liste.append(zusaetzliches_element)
print(liste)
```

```python
# falls mehrere Element hinzugefügt werden sollen
liste.extend(zusaetzliche_elemente_l)  # Parameter muss iterierbar sein, alternativ kann auch +-Operator verwendet werden.
print(liste)
liste_2 = liste + zusaetzliche_elemente_l
print(liste_2)
```

```python
# falls mittels append ein iterierbarer Parameter übergeben wird
liste.append(zusaetzliche_elemente_l)  # iterierbares Element wird als einzelnes Element hinzugefügt
print(liste)
print(liste[-1], liste[-1][-1])
```

```python
# i-te Element zurückgeben und entfernen
letzte_element = liste.pop()  # ohne Parameter wird das letzte Element entfernt
print(letzte_element)
print(liste)
```

```python
i_te_element = liste.pop(2)  # i-te Element entfernen
print(i_te_element)
print(liste)
```

```python
# entfernen eines Elementes (Wertes) ohne Kenntnis des entsprechenden Indexes
print(liste)
liste.remove(34)  # es wird nur jeweils ein Element entfernt. Falls Wert mehrfach vorhanden ist, muss remove() mehrmals ausgeführt werden.
print(liste)
```

```python
# zählen von Werten in einer Liste/Tupel
anzahl_liste = liste.count(345.34)
anzahl_tupel = tupel.count(4.5)
print(anzahl_liste, liste)
print(anzahl_tupel, tupel)
# ist entsprechendes Element enthalten?
print('Element in Liste? ', 345.34 in liste)
print('Element in Tupel? ', 4.5 in tupel)
```

```python
# Position eines Elementes innerhalb einer Liste ermitteln (es wird nur der Index des ersten Elements zurückgegeben)
index_liste = liste.index(55)
index_tupel = tupel.index(451)
print('Index in Liste: ', index_liste)
print('Index in Tupel: ', index_tupel)
# optional können noch Start- und Endindex der Suche übergeben werden, z.B. liste.index(55, 1, 3)
```

```python
# einfügen eines Elementes an eine beliebige Stelle innerhalb einer Liste (nicht am Schluss, sonst append())
print(liste)
liste.insert(2, 'neues Element')
print(liste)
```

```python
# Mehrfachzuweisungen, Packing und Unpacking
erste, *weitere = 34, 44, 6743, 89
print(erste)
print(weitere)  # wird in ein Tupel gepackt
```

```python
liste = [100, 3.9, 451, 55, 34]
# sortieren von Listen
print('Die unsortierte Liste: ', liste)
liste.sort()  # sort() liefert nicht die sortierte Liste zurück! Rückgabewert ist None
print('Die sortierte Liste: ', liste)
# Alternative
liste_sortiert = sorted(liste)
print('Die sortierte Liste: ', liste_sortiert)
# die Sortierreihenfolge kann auch invertriert werden
liste.sort(reverse=True)  # oder auch sorted(liste, reverse=True)
print('Die Liste umgekehrt sortiert: ', liste)
```
