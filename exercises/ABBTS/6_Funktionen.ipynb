# 6. Funktionen
Ganz allgemein gesehen, stellt eine Funktion ein Strukturierungselement in einer Programmiersprache dar. Damit kann eine Menge von Anweisungen gruppiert, bzw. zusammengefasst werden. Ebenso wird es dadurch möglich Codeinhalt mehrfach an verschiedenen Orten zu nutzen. <br>
Die Benutzung von Funktionen erhöht ausserdem die Verständlichkeit und Qualität eines Programms. Weiter wird die Wartbarkeit einfacher.

````python
# Funktionsdefinition mit Schlüsselwort def

# def funktionsname(Übergabeparameter):
#   Anweisung(en)
#   return Rückgabeparameter

# Übergabeparameter ist mit einem Standardwert = 10 versehen,
# d.h. beim Funktionsaufruf ohne Übergabeparameter wird bis 10 gezählt und summiert.

def zaehle_summiere(start=1, ende=10):
    """
    Funktion zählt von Start bis zum Ende (inklusiv).
    Zusätzlich enthält der Rückgabeparameter die Summe von 
    Start + (Start+1) + (Start+2) + ... + Ende.
    """
    zahlen = range(start, ende+1)
    for i in zahlen:
        print(i)

    # Aufsummieren 1 bis Endzahl
    summe = sum(zahlen)
    print('Summe lautet: ', summe)
    # Rückgabewerte
    return zahlen, summe
````

````python
zahlen, summe = zaehle_summiere(5, 10)
````

````python
help(zaehle_summiere)
````

````python
# keine Übergabeparameter gesetzt, d.h. Funktion übernimmt die Standardwerte 1, 10
print('Funktion ohne Übergabeparameter aufrufen')
zaehle_summiere()
````

````python
# ein Übergabeparameter gesetzt, d.h. Funktion übernimmt 5 - 10
print('Funktion mit nur einem Übergabeparameter aufrufen')
zaehle_summiere(5)
````

````python
# falls nun aber der Startwert als Standard gesetzt werden will und der Endwert angepasst,
# muss die folgende Definition verwendet werden
print('Funktion mit nur Endwert aufrufen -> Schlüsselwortarameter')
zaehle_summiere(ende=5)  # Schlüsselwortparameter
````

## 6.1 Lokale und globale Variablen
Variablen sind standardmässig lokal, d.h. sie sind nur innerhalb der Funktion sichtbar, in welcher sie definiert worden sind. Es ist auch möglich, innerhalb einer Funktion (schreibend) auf globale Variablen zuzugreifen.

````python
# Beispiel globale Variablen beschreiben innerhalb einer Funktion

def funktion():
    global a  # Explizit die globale Variable a deklariert
    print('a = ', a)
    a += 1
    print('a = ', a)
````

````python
a = 5

funktion()
````

## 6.2 Variable Anzahl Parameter

````python
# Variable Anzahl Übergabeparameter
# def funktion(plicht, *freiwillig)

def summiere_v01(erster_wert, *weitere_werte):
    summe = erster_wert + sum(weitere_werte)
    anzahl_werte = 1 + len(weitere_werte)

    return summe, anzahl_werte
````

````python
summe, anzahl_summanden = summiere_v01(1, 2, 3.4, 4.2)
print('Summe lautet: ', summe)
print('Anzahl Summanden: ', anzahl_summanden)
````

````python
liste = [1, 2, 3.4, 4.2]
# mit * im Funktionsaufruf werden die Elemente einer Liste oder Tupels entpackt (vereinzelt).
(summe, anzahl_summanden) = summiere_v01(*liste)
print('Summe lautet: ', summe)
print('Anzahl Summanden: ', anzahl_summanden)
````

## 6.3 Beliebige Anzahl Schlüsselwortparameter
Es ist auch möglich eine beliebige Anzahl Schlüsselwortparameter zu übergeben. Diese werden in ein Dictionary gepackt.

````python
# Beliebige Anzahl Schlüsselwortparameter (z.B. start=5, ende=10)
def summiere_v02(**werte):
    print(werte)
    print(type(werte))
    # Schlüsselwörter des Dictionary zurückgeben
    werte_keys = werte.keys()
    werte_values = werte.values()
    print(werte_keys, werte_values)

    return
````

````python
summiere_v02(erster=34, weitere=43)
````

## 6.5 lambda
Eine anonyme Funktion oder Lambda-Funktion ist eine Funktion, die über keinen Namen verfügt. Eine solche Funktion kann folglich nur über Verweise angesprochen werden. Sie können eine beliebige Anzahl Parameter haben, führen einen Ausdruck aus und liefern den dieses Ausdrucks als Rückgabewert zurück. <br>
Ein einfaches Beispiel:
```python
lambda x: x + 42
```
Beim obigen Beispiel handelt es sich um eine Funktion mit einem Argument x, die die Summe von x + 42 zurückgibt. Der allgemeine Syntax einer Lambda-Funktion sieht folgendermassen aus:
```python
lambda Argumentenliste: Ausdruck
```
Die Argumentenliste besteht aus einer durch Komma getrennten Liste von Argumenten, und der nach dem Doppelpunkt stehende Ausdruck ist ein beliebiger Ausdruck, der diese Argumente benutzt. Was macht man nun damit?

````python
f42 = lambda x: x + 42  # Eine lambda-Funktion einer Variablen zuordnen
f42(8)
````

Nun noch eine sinnvollere Anwendung solcher Funktionen. Es soll eine Funktion mit dem Namen "anwenden" implementiert werden, die als erstes Argument eine Funktion und als zweites Argument eine Liste erwartet. Die Funktion "anwenden" wendet auf jedes Element dieser Liste die übergebene Funktion an.

````python
def anwenden(f, liste):
    ergebnis = []
    for element in liste:
        ergebnis.append(f(element))
    return ergebnis
````

````python
anwenden(f42, range(10))
````

Diese Art der Implementation erzeugt eine sogenannte Wegwerffunktion f42, da diese nur einmal beim Funktionsaufruf von "anwenden" benötigt wird. Aus diesem Grund wäre es viel eleganter wenn diese Funktion direkt an die Funktion "anwenden" übergeben werden kann, also ohne den Umweg mit der Namensgebung "f42". Dies ist mit der Lambda-Notation möglich:

````python
anwenden(lambda x: x+42, range(10))
````

````python
for i in [17, 22, 50]:  # via for drei unterschiedliche Funktionen erzeugt
    ergebnis = anwenden(lambda x: x + i, range(10))
    print(ergebnis)
````
