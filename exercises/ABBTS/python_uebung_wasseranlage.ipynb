Aufgabe 6

Eine Anlage enthält 2 Sensoren, S1 und S2. Speichere die Messwerte von S1 und S2 in einen Dictionary mit dem Namen Anlage:

Keys: 

    S1_Messwerte
    S1_Fehlercodes
    S1_Grenzen

    S2_Messwerte
    S2_Fehlercodes
    S2_Grenzen

Messwerte Sensor 1: 135, 150, 142, 160, 150, 155
Messwerte Sensor 2: 122, 131, 111, 990, 11, 15

Grenzen Sensor 1: 50,150
Grenzen Sensor 2: 10,250

Fehlercodes 1: 101,102,101,103
Fehlercodes 2: 101,999,999,999

-Gib alle Messerte von Sensor 1 aus
-Bestimme den Maximalwert von Sensor 2
-Prüfe für Sensor 1 welche Messwerte über dem Grenzwert liegen
-Zähle wie oft der Fehlercode 999  bei Sensor 2 vorkommt

-Erstelle eine Schlaufe welche alle Messerte durchläuft, prüft ob die Werte innherhalb der Grenzen liegen und wenn die Werte ausserhalb liegen, ausgegeben werden.

```python
S1_Messwerte = [135, 150, 142, 160, 150, 155]
S2_Messwerte = [122, 131, 111, 990, 11, 15]

limitS1 = (50, 150)
limitS2 = (10, 250)

fehlercodesS1 = [101, 102, 101, 103]
fehlercodesS2 = [101, 999, 999, 999]

wasseranlage = {
    'S1 Messwerte': S1_Messwerte, 
    'S1 Fehlercodes': fehlercodesS1, 
    'S1 Grenzen': limitS1, 
    'S2 Messwerte': S2_Messwerte,
    'S2 Fehlercodes': fehlercodesS2,
    'S2 Grenzen': limitS2
    }

# Alle Messwerte von s1 ausgeben
print(wasseranlage['S1 Messwerte'])

# Maximalwert von S2 bestimmen 
print(max(wasseranlage['S2 Messwerte']))

# Prüfen welche Messwerte über dem Grenzwert liegen
for i in wasseranlage['S1 Messwerte']:
    if i > max(wasseranlage['S1 Grenzen']):
        print(f'Warnung grenzwert überschritten mit wert: {i}')
    
# Zählen wie oft der Messwert '999' bei Sensor 2 vorkommt
# feedback: mit .count(999) noch eleganter
count = 0
for i in wasseranlage['S2 Fehlercodes']:
    if i == 999:
        count += 1

print(f'Fehlermeldung mit ID: 999 ist {count} mal aufgetretten.')

# Prüfe alle Messergebnisse und gibt alle ausserhalb aus. Muss nicht als funktion def werden aber finde es schöner so
def limit_check(lo, hi, werte):
    outOfSpec = []
    for i in werte:
        if i > hi or i < lo:
            outOfSpec.append(i)
    return outOfSpec

print(f'The following values were out of spec for S1: {limit_check(min(wasseranlage['S1 Grenzen']), max(wasseranlage['S1 Grenzen']), wasseranlage['S1 Messwerte'])}')

print(f'The following values were out of spec for S2: {limit_check(min(wasseranlage['S2 Grenzen']), max(wasseranlage['S2 Grenzen']), wasseranlage['S2 Messwerte'])}')
```

```python
# Verbesserte version mit feedback von claude:

S1_Messwerte = [135, 150, 142, 160, 150, 155]
S2_Messwerte = [122, 131, 111, 990, 11, 15]

limitS1 = (50, 150)
limitS2 = (10, 250)

fehlercodesS1 = [101, 102, 101, 103]
fehlercodesS2 = [101, 999, 999, 999]

wasseranlage = {
    "S1 Messwerte": S1_Messwerte,
    "S1 Fehlercodes": fehlercodesS1,
    "S1 Grenzen": limitS1,
    "S2 Messwerte": S2_Messwerte,
    "S2 Fehlercodes": fehlercodesS2,
    "S2 Grenzen": limitS2
}

# Alle Messwerte von S1
print(wasseranlage["S1 Messwerte"])

# Maximalwert von S2
print(max(wasseranlage["S2 Messwerte"]))

# Grenzwertüberschreitungen S1
lo, hi = wasseranlage["S1 Grenzen"]
for w in wasseranlage["S1 Messwerte"]:
    if w > hi:
        print(f"Warnung: Grenzwert überschritten mit Wert {w}")

# Fehlercode 999 zählen
count = wasseranlage["S2 Fehlercodes"].count(999)
print(f"Fehlercode 999 ist {count} mal aufgetreten.")

# Funktion zur Grenzprüfung
def limit_check(lo, hi, werte):
    return [w for w in werte if w < lo or w > hi]
        #   ^      ^    ^^^^^^^^^^^^^^^^^^^^^^^^^                    <-- Erklärung was was ist
        #   |      |              |
        #   |      |              └→ Bedingung (Filter)
        #   |      └→ Quelle der Werte (Iteration)
        #   └→ Was in die neue Liste geschrieben wird


print(f"S1 out of spec: {limit_check(*wasseranlage['S1 Grenzen'], wasseranlage['S1 Messwerte'])}") # * zum automatischen enpaken des tuppels
print(f"S2 out of spec: {limit_check(*wasseranlage['S2 Grenzen'], wasseranlage['S2 Messwerte'])}")
```
