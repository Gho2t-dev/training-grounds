### 2.3.1 Dictionaries
Ein Dictionary besteht aus Schlüssel-Werte-Paaren. Jedem Schlüssel-Objekt ist ein Werte-Objekt zugeordnet. Daher auch die Bezeichnung *Mapping* für diese Kategorie von Datentypen. Dictionaries gehören zu den wichtigsten Datentypen in Python. Wie Listen können Dictionaries leicht verändert werden, ausserdem können sie zur Laufzeit beliebig wachsen oder schrumpfen.

```python
# Die Schlüssel-Werte-Paare werden mittels , voneinander getrennt
d = {'Reglertyp': 'PID', 'Kp': 2.45, 'Ti': 0.34, 'Td': 1.2}
print(d)
e = {(2.5, 3.0): 'Student X', (4.0, 3.0): 'Student Y', (1.5, 5.0): 'Student Z'}
print(e)
```

Als Schlüssel sind neben Strings auch Integer, Floats oder Tupels zulässig.

Im Unterschied zu Listen werden zur Indizierung keine Zahlen, sondern die Schlüssel benutzt. Das heisst für Dictionaries ist keine Ordnung definiert. Python kann die Schlüssel-Werte-Paare beliebig ausgeben.

```python
d = {'Reglertyp': 'PID', 'Kp': 2.45, 'Ti': 0.34, 'Td': 1.2}
print(d['Reglertyp'])
```

#### 2.3.1.1 Zugriffe auf Dictionaries
Mit dem Schlüsselwort *in* kann auf die Schlüssel eines Dictionaries zugegriffen werden. Dieses liefert *True* zurück, falls der Schlüssel existiert und *False*, falls der Schlüssel nicht existiert. Alternativ kann mit *get* auf die Schlüssel zugegriffen werden. Diese erlaubt im Gegensatz zur ersten Variante die Definition eines Default-Wertes, falls der Schlüssel nicht existiert.

```python
d = {'Reglertyp': 'PI', 'Kp': 2.45, 'Ti': 0.34, 'Td': 2.34}
print('Ti' in d)
print('Td' in d)
d.get('Td', 'nicht vorhanden')  # Default-Wert: 'nicht vorhanden'
```

### 2.3.2 Methoden auf Dictionaries
Der Begriff Methoden stammt aus der objektorientierten Programmierung (OOP). Dazu folgt später ein eigenes Kapitel. Für das Verständnis von den folgenden Funktionalitäten braucht es dieses Wissen zur OOP noch nicht.
1. *copy()*
2. *clear()*
3. *fromkeys()*
4. *items()*
5. *keys()*
6. *pop()*
7. *popitem()*
8. *setdefault()*
9. *update()*

```python
''' Item hinzufügen hiermit oder einfach mit .update({'color': 'red'})'''
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
thisdict["color"] = "red"
print(thisdict)

# copy()
''' erzeugt eine Kopie eines Dictionaries. '''
d = {'Reglertyp': 'PI', 'Kp': 2.45, 'Ti': 0.34}
e = d.copy()  # Kopiert ein Dictionary, es wird ein neues Objekt erzeugt
print('d lautet: ', d)
print('e lautet: ', e)
print('Sind die beiden Objekte identisch? ', id(d) == id(e))  # ist ein neues Objekt!
e['Ti'] = 0.5
print('Nach der Änderung von e[Ti]')
print('d lautet: ', d)
print('e lautet: ', e)

# im Unterschied zu einer Zuweisung --> ist keine Kopie!
f = d  # f ist dasselbe Objekt wie d
print('Sind die beiden Objekte identisch? ', id(d) == id(f))
f['Ti'] = 0.5  # Konsequenz? verändert auch d
print('Nach der Änderung von f[Ti]')
print('d lautet: ', d)
print('f lautet: ', f)
```

```python
# clear()
''' löscht die Elemente eines Dictionaries. '''
d = {'Reglertyp': 'PI', 'Kp': 2.45, 'Ti': 0.34}
e.clear()  # Löscht die Schlüssel-Werte-Paare, das Dictionary aber nicht
print(d, e)
print(type(e))
```

```python
# dict.fromkeys(S,[,v])
''' liefert ein Dictionary zurück mit den in S definierten Schlüsseln und einem optionalen,
    aber für alle Werte identischen Wert. '''
bezeichnung = ('Reglertyp', 'Kp', 'Ti', 'Td')
d = dict.fromkeys(bezeichnung)
print(d)
e = dict.fromkeys(bezeichnung, 2)
print(e)
```

```python
# items()
''' liefert eine View der Schlüssel-Werte-Paare eines Dictionaries zurück. '''
d = {'Reglertyp': 'PI', 'Kp': 2.45, 'Ti': 0.34}
d_view = d.items()
print(d_view)
```

```python
# keys()
''' liefert eine View der Schlüssel eines Dictionaries zurück. '''
d = {'Reglertyp': 'PI', 'Kp': 2.45, 'Ti': 0.34}
d_view = d.keys()
print(d_view)
```

```python
# pop(k[,x])
''' entfernt das Schlüssel-Werte-Paar (k) aus dem Dictionary. Entspricht k keinem Schlüssel,
    wird der optionale Parameter x zurückgegeben. '''
d = {'Reglertyp': 'PI', 'Kp': 2.45, 'Ti': 0.34}
print(d.pop('Kp'))
print(d)
print(d.pop('Kp', 'nicht vorhanden'))
```

```python
# popitem()
''' wird ohne Parameter aufgerufen und liefert ein beliebiges Schlüssel-Werte-Paar des
    Dictionary's zurück. Dieses Paar wird dann ebenfalls entfernt. '''
d = {'Reglertyp': 'PI', 'Kp': 2.45, 'Ti': 0.34}
print(d.popitem())
print(d)
```

```python
# setdefault(k[,x])
''' setzt den Wert des Schlüssels k auf x, falls k noch nicht vorhanden ist. Ist der optionale
    Parameter x nicht definiert, wird der Wert auf 'None' gesetzt. Falls der Schlüssel k
    bereits vorhanden ist, wird der dazugehörige Wert nicht verändert. '''
d = {'Reglertyp': 'PI', 'Kp': 2.45, 'Ti': 0.34}
d.setdefault('Td', 2.4)
print(d)
d.setdefault('Beschreibung')
print(d)
# falls der Schlüssel bereits vorhanden ist, erfolgt keine Anpassung des Wertes
d.setdefault('Kp', 0.5)
print(d)
```

```python
# update(dict)
''' fügt dem ursprünglichen Dictionary ein zweites hinzu und überschreibt ggf. die Werte
    von bereits vorhandenen Schlüsseln. '''
d = {'Reglertyp': 'PI', 'Kp': 2.45, 'Ti': 0.34}
d2 = {'Reglertyp': 'PID', 'Td': 2.0}
d.update(d2)
print(d)
```

### 2.3.3 Aus Listen Dictionaries erzeugen (zip-Funktion)
Diese sehr nützliche Funktion kann dazu genutzt werden, aus Listen Dictionaries zu erzeugen.

```python
studenten = ['Student X', 'Student Y', 'Student Z', 'Student W']
noten = [5.2, 4.3, 3.5, 5.7]
studenten_noten = dict(zip(studenten, noten))
print(studenten_noten)
```

```python
netzwerk = {'R': 1e3, 'L': 10e-3, 'C': 6.8e-6 }
for key in netzwerk:
    value = netzwerk[key]
    print('Bauteil {0} mit dem Wert {1}'.format(key, value))
```
