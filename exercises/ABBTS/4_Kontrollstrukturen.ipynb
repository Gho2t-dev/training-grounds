# 4. Kontrollstrukturen
Anhand von Kontrollstrukturen kann man einen rein linearen Programmablauf beeinflussen und kontrollieren. Dazu gehören Verzweigungen und Schleifen.

## 4.1 Verzweigungen


```python
tastatur_eingabe = eval(input('Bitte einen beliebigen Datentyp eingeben: '))

if isinstance(tastatur_eingabe, int):
    print('Sie haben eine Ganzzahl eingegeben')
elif isinstance(tastatur_eingabe, float):
    print('Sie haben eine Fliesskommazahl eingegeben')
elif isinstance(tastatur_eingabe, list):
    print('Sie haben eine Liste eingegeben')
elif isinstance(tastatur_eingabe, tuple):
    print('Sie haben ein Tuple eingegeben')
elif isinstance(tastatur_eingabe, dict):
    print('Sie haben ein Dictionary eingegeben')
else:
    print('Dieser Datentyp wurde nicht erkannt')
```

### 4.1.1 Vergleichsoperatoren

### 4.1.2 Zusammengesetzte Bedingungen
Bedingungen können mittels logischen Operatoren *and*, *or* und *not* zusammengesetzt werden.

```python
a = 42
if 0 <= a and a < 100:
    print('Die Zahl liegt zwischen 0 und 100')
else:
    print('Die Zahl ist negativ oder grösser 100')
```

## 4.2 Schleifen
Schleifen werden benötigt um einen bestimmten Codeabschnitt wiederholt auszuführen. Es existieren die beiden Schleifen *while* und *for*.

### 4.2.1 while-Schleife
Bei der while-Schleife von Python handelt es sich um eine sogenannte kopfgesteuerte Schleife. D.h. bevor der Schleifenrumpf ausgeführt wird, wird die Bedingung im Schleifenkopf geprüft. <br>
Manchmal würde man aber lieber zuerst den Schleifenrumpf ausführen und danach die Bedingung im Schleifenkopf prüfen, eine sogenannte fussgesteuerte Schleife. Python bietet keine solche Schleife. Man kann diese jedoch mit einem *while True* in Verbindung mit einer *break*-Anweisung am Ende des Schleifenkörpers erzeugen.

```python
i = 1
while i <= 5:
    print(i, i**2)
    i += 1
```

### 4.2.2 break und continue
Für Schleifen existieren zwei wichtige Anweisungen. *break* für den vorzeitigen Abbruch und *continue*, um einen Durchlauf zu beenden und mit dem nächsten Durchlauf weiterzumachen.<br>
Stösst der Programmablauf auf eine break-Anweisung, wird die Schleife unmittelbar abgebrochen. Das Programm wird mit der ersten Anweisung nach der Schleife fortgesetzt. Bei geschachtelten Schleifen wird mit break nur die innerste Schleife abgebrochen. Trifft der Programmablauf auf eine continue-Anweisung, wird der Schleifendurchlauf abgebrochen und der Programmablauf kehrt zum Schleifenkopf zurück, wo geprüft wird, ob die Bedingung für einen weiteren Durchlauf erfüllt ist. <br>
Das nachfolgende Beispiel soll die unterschiedlichen Funktionalitäten dieser beiden Anweisungen veranschaulichen.

```python
liste = eval(input('Liste mit positiven Zahlen eingeben:'))
n = len(liste)
i = 0
zahlen = []
while i < n:
    aktuelle_zahl = liste[i]
    i += 1
    if aktuelle_zahl in zahlen:  # gleiche Zahlen werden ignoriert
        continue
    if aktuelle_zahl < 0:
        print('Abbruch: Negative Zahl gefunden')
        break
    zahlen.append(aktuelle_zahl)

print(zahlen)
```

### 4.2.2 for-Schleife
Bei der for-Schleife von Python handelt es sich um ein Sprachkonstrukt, mit dessen Hilfe nacheinander beispielsweise die Elemente einer Menge oder Liste bearbeitet werden können. Dazu werden sie einer Variablen zugewiesen. Wie die while-Schleife besitzt die for-Schleife ebenfalls einen else-Block.

```python
studenten = ['Student X', 'Student Y', 'Student Z', 'Student W']
noten = [5.2, 4.3, 3.5, 5.7]
studenten_noten = dict(zip(studenten, noten))

for student in studenten_noten:  # Iteration über die Schlüssel eines Dictionaries
    if studenten_noten[student] < 4.0:
        print(student, 'leider ungenügend')
        continue
    print(student, studenten_noten[student])
```

```python
name = 'Felix Müller'

for zeichen in name:  # Iteration über die Zeichen eines Strings
    print(zeichen)
```

```python
for i in range(0, 8, 1):  # Start, Ende, Schrittweite
    print(i)

print('Ende')
```

```python
anzahl = 0
for i in range(0, 8, 1):
    for j in range(0, 8, 1):
        print('ij: ', i, j)
        anzahl += 1

print('Anzahl: ', anzahl)
```

```python
zahl = 2

if zahl != 2:
    print('gleich')
else:
    print('ungleich')
```
