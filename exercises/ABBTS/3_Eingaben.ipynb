# 3. Eingabe mittels input
Eingaben können über verschiedene Wege erfolgen. Ein häufiger Weg ist via Tastatur. Der Datentyp der Funktion *input* ist immer ein String.

```python
tastatur_eingabe = input('Bitte die Eingabe: ')
print('Die Eingabe lautet: ', tastatur_eingabe)
print('Der Datentyp der Eingabe lautet: ', type(tastatur_eingabe))
```

Soll der Datentyp der Eingabe ausgewertet werden, so kann dies mit der Funktion *eval* erfolgen.

```python
tastatur_eingabe = eval(input('Bitte eine Fliesskommazahl eingeben: '))
print('Der Datentyp der Eingabe lautet: ', type(tastatur_eingabe))
```

```python
tastatur_eingabe = eval(input('Bitte eine Liste eingeben: '))
print('Der Datentyp der Eingabe lautet: ', type(tastatur_eingabe))
```

```python
tastatur_eingabe = eval(input('Bitte ein Dictionary eingeben: '))
print('Der Datentyp der Eingabe lautet: ', type(tastatur_eingabe))
```
